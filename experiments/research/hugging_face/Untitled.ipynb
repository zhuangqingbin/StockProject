```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")
```

```python
?pipeline
```

```python
classifier(
    ["I've been waiting for a HuggingFace course my whole life.", "I feeling better now", "I feeling worse now", "我的心情一般", "我的心情晴转多云了", "我的心情很好"]
)
```

```python
zs_classifier = pipeline("zero-shot-classification")
zs_classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
generator = pipeline("text-generation")
generator("In this course, we will teach you how to", num_return_sequences = 4, max_length = 100)
```
