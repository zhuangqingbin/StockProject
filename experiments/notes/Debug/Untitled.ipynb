```python
def agg_func(df):
    no_click = df[df['click_per']==0].cnt.sum()
    click_eq_imp = df[df['click_per']==0.5].cnt.sum()
    click_lt_imp = df[(df['click_per']<0.5) & (df['click_per'] != 0)].cnt.sum()
    all_cnt = df.cnt.sum()
    return pd.Series([no_click * 1.0 / all_cnt, click_eq_imp * 1.0 / all_cnt,
                click_lt_imp * 1.0 / all_cnt], ['no_click_percent', 'click_eq_imp_percent', 'click_lt_imp_percent'])
```

```python
df.groupby(['dt', 'hour']).apply(agg_func).reset_index()
```

```python
df.groupby(['dt', 'hour']).apply(agg_func).reset_index().to_csv("tmp.csv")
```
