```python
slot_lst, weight_lst = [], []
for _file in os.listdir(Dir_path):
    if not _file.endswith('csv'):
        continue
    df = pd.read_csv(os.path.join(Dir_path, _file))
    weight = df.sort_values('Step')['Value'].values[-1]
    slot = re.findall('get_slot_(.*?)_category', _file)[0]
    
    slot_lst.append(slot)
    weight_lst.append(weight)
    
df_summary = pd.DataFrame({'slot': slot_lst, 'weight': weight_lst})
```
