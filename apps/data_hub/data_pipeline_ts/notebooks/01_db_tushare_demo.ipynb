# DB + TuShare Demo

这个 notebook 保持最小化：
- 数据库连接字符串直接在 notebook 里改
- TuShare 通过环境变量 `TUSHARE_TOKEN` 创建 `pro`
- 适合直接在 IDE 里打开后逐格运行


```python
import os

import pandas as pd
import tushare as ts
from sqlalchemy import create_engine
```

```python
# 把这里改成你自己的数据库连接串
DB_URL = "mysql+pymysql://user:password@127.0.0.1:3306/your_database?charset=utf8mb4"

engine = create_engine(DB_URL, pool_pre_ping=True)
QUERY_SQL = "SELECT DATABASE() AS current_database"
pd.read_sql_query(QUERY_SQL, con=engine)
```

```python
# 你也可以把查询 SQL 改成任意表
QUERY_SQL = "SELECT DATABASE() AS current_database"
pd.read_sql_query(QUERY_SQL, con=engine)
```

```python
if not os.getenv("TUSHARE_TOKEN"):
    raise ValueError("Please export TUSHARE_TOKEN before running this cell.")

pro = ts.pro_api(os.getenv("TUSHARE_TOKEN"))
pro.stock_basic(exchange="", list_status="L", fields="ts_code,symbol,name,area,industry,list_date").head()
```
