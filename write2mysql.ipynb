```python
start, end = '20250808', '20250816'
```

```python
dates = pd.date_range(start=start, end=end, freq='D')
date_list = dates.strftime('%Y%m%d').tolist()  # 转为字符串列表
print(date_list[:5])  # 查看前5个日期
print(f"共计 {len(date_list)} 天")

stock_daily = StockDailyFetch()
for date in date_list:
    time.sleep(0.5)
    stock_daily.fetch_data(f"{date} 数据", trade_date = date, adj = 'qfq')
    df = stock_daily.data
    if len(df) > 0:
        print("==> 写入数据")
        # df['trade_date'] = pd.to_datetime(df['trade_date'])
        try:
            df.to_sql('daily_kline', engine, if_exists='append', index=False, method='multi')
        except IntegrityError as e:
            print("    ==> 插入失败：主键重复，部分记录已存在")
        except Exception as e:
            print("    ==> 其他错误：", e)
    else:
        print("==> 非交易日")
    print('\n')
```
